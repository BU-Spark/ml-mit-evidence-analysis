## Visualize prediction

```python
import os

def get_image_paths(folder_path):
    # Get all file names in the folder
    file_names = os.listdir(folder_path)
    # Full paths of files
    full_paths = [os.path.join(folder_path, file_name) for file_name in file_names]
    # Filter out non-image files, adjust the extensions as needed
    image_paths = [path for path in full_paths if path.lower().endswith(('.png', '.jpg', '.jpeg', '.tiff', '.bmp', '.gif'))]
    return image_paths

# Use the function like this:
test_folder = 'test'
result_folder = 'results'
test_img_paths = get_image_paths(test_folder)
result_img_paths = get_image_paths(result_folder)
print(test_img_paths)
print(result_img_paths)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

# Assuming test_img_paths and result_img_paths are lists containing the paths to the images.
for i in range(len(test_img_paths)):
    test_img = Image.open(test_img_paths[i])
    result_img = Image.open(result_img_paths[i])

    plt.figure(figsize=(10,5))  # Set the figure size as per your requirement

    # Display test image
    plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st subplot
    plt.imshow(test_img)
    plt.title('Test Image')
    plt.axis('off')  # Hide the axis

    # Display result image
    plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd subplot
    plt.imshow(result_img)
    plt.title('Result Image')
    plt.axis('off')  # Hide the axis

    plt.show()
```
